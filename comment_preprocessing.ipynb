```python
import re
import string
```

```python
vietnamese_stopwords = set([
    "và", "là", "của", "có", "cho", "với", "một", "các", "những", "được",
    "trong", "khi", "đã", "thì", "lại", "này", "kia", "đó", "nhiều", "rất",
    "ra", "vào", "từ", "vẫn", "nên", "đến", "nơi", "vì", "bị", "ở", "sẽ",
    "đang", "vừa", "cùng", "như", "hay", "nếu", "vậy", "thế", "rằng", "về",
    "vẫn", "đều", "thôi", "đi", "thật", "chỉ"
])

replace_dict = {
    "ko": "không",
    "hok": "không",
    "k": "không",
    "kg": "không",
    "khong": "không",
    "dc": "được",
    "đc": "được",
    "dk": "được",
    "ok": "được",
    "đk": "được",
    "vs": "với",
    "ms": "mới",
    "j": "gì",
    "v": "vậy",
    "k0": "không",
    "k0g": "không",
    "kp": "không phải",
    "bt": "bình thường"
}

def normalize_teencode(text):
    for k, v in replace_dict.items():
        text = re.sub(rf"\b{k}\b", v, text)
    return text

def simple_tokenize(text):
    return re.findall(r'\w+', text)

def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    text = text.lower()
    text = normalize_teencode(text)
    text = re.sub(r"http\S+|www\S+|https\S+", '', text)
    text = re.sub(r'[^\w\sàáạảãâầấậẩẫăằắặẳẵèéẹẻẽêềếệểễìíịỉĩòóọỏõôồốộổỗơờớợởỡùúụủũưừứựửữỳýỵỷỹđ]', ' ', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(f"[{re.escape(string.punctuation)}]", " ", text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    tokens = simple_tokenize(text)
    tokens = [t for t in tokens if t not in vietnamese_stopwords]
    
    return " ".join(tokens)
```

```python
import json

with open("restaurants_reviews.json", "r", encoding="utf-8") as f:
    reviews = json.load(f)

texts = []
for r in reviews:
    if "review_text" in r:
        texts.append(clean_text(r["review_text"]))

with open("reviews_text_preprocessing.json", "w", encoding="utf-8") as f:
    json.dump(texts, f, ensure_ascii=False, indent=4)

print("✅ Đã lưu file reviews_text_only.json thành công!")
```
