```python
import json

with open("restaurants_reviews.json", "r", encoding="utf-8") as f:
    reviews = json.load(f)

with open("user_list.json", "r", encoding="utf-8") as f:
    users = json.load(f)

user_map = {user["username"]: user["user_id"] for user in users}

review_summary = []

for review in reviews:
    username = review["username"].strip()
    user_id = user_map.get(username)  
    if user_id is not None:
        try:
            score = float(review["score_text"])
        except (ValueError, TypeError):
            score = None
        review_summary.append({
            "review_id": review["review_id"],
            "user_id": user_id,
            "res_id": review["res_id"],
            "rating": score
        })
    else:
        print(f"⚠️ Không tìm thấy user_id cho username: {username}")

with open("review_summary.json", "w", encoding="utf-8") as f:
    json.dump(review_summary, f, ensure_ascii=False, indent=4)

print("✅ Đã tạo file review_summary.json thành công!")
```
