```python
import json
with open("restaurants_reviews.json", "r", encoding="utf-8") as f:
    reviews = json.load(f)

users = {} 
user_list = []  # danh sách user để lưu ra file, DB,...
 
next_id = 1

for review in reviews:
    username = review["username"].strip()
    if username not in users:
        users[username] = next_id
        user_list.append({
            "user_id": next_id,
            "username": username
        })
        next_id += 1

print(user_list)
```
